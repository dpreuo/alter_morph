```python
import numpy as np 
import matplotlib.pyplot as plt
import matplotlib
```

```python
def T_ij(theta,t1=1,t2=0.5):
    theta = np.array(theta)
    T = np.zeros((2,2,*theta.shape))
    T[0,0] = (t1-t2)*np.cos(theta)**2 + t2
    T[1,1] = (t1-t2)*np.sin(theta)**2 + t2
    T[0,1] = T[1,0] = (t1-t2)*np.sin(theta) * np.cos(theta)
    
    return T
```

```python

fig, ax = plt.subplots(1,1,subplot_kw={'projection': 'polar'})

angles = np.linspace(0,2*np.pi,100)

ax.plot(angles, T_ij(angles)[0,0],label='Txx')
ax.plot(angles, T_ij(angles)[1,1],label='Tyy')
ax.plot(angles, np.abs(T_ij(angles)[0,1]),label='|Txy|')

ax.legend()
```
