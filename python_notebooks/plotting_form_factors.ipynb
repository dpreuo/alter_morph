```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib

from alter_morph.hamiltonians import _hopping_matrix
```

```python
def T_ij(t1, t2, theta):
    theta = np.array(theta)
    T = np.zeros((2, 2, *theta.shape))
    T[0, 0] = (t1 - t2) * np.cos(theta) ** 2 + t2
    T[1, 1] = (t1 - t2) * np.sin(theta) ** 2 + t2
    T[0, 1] = T[1, 0] = (t1 - t2) * np.sin(theta) * np.cos(theta)

    return T
```

```python
fig, ax = plt.subplots(1, 3, subplot_kw={"projection": "polar"}, figsize=(10, 3))

angles = np.linspace(0, 2 * np.pi, 100)

t1 = 1
t2 = 0.2

matrix_1 = -_hopping_matrix(t1, t2, angles, n=1)
matrix_2 = -_hopping_matrix(t1, t2, angles, n=2)
matrix_3 = -_hopping_matrix(t1, t2, angles, n=3)

ax[0].plot(angles, matrix_1[0, 0], label="Txx")
ax[0].plot(angles, matrix_1[1, 1], label="Tyy")
ax[0].plot(angles, matrix_1[0, 1], label="|Txy|")
ax[0].plot(angles, matrix_1[1, 0], label="|Tyx|")

ax[1].plot(angles, matrix_2[0, 0], label="Txx")
ax[1].plot(angles, matrix_2[1, 1], label="Tyy")
ax[1].plot(angles, matrix_2[0, 1], label="|Txy|")
ax[1].plot(angles, matrix_2[1, 0], label="|Tyx|")

ax[2].plot(angles, matrix_3[0, 0], label="Txx")
ax[2].plot(angles, matrix_3[1, 1], label="Tyy")
ax[2].plot(angles, matrix_3[0, 1], label="|Txy|")
ax[2].plot(angles, matrix_3[1, 0], label="|Tyx|")


# ax[1].legend()
```
