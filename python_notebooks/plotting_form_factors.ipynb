```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib

from alter_morph.hamiltonians import _hopping_matrix
```

```python
def T_ij(t1, t2, theta):
    theta = np.array(theta)
    T = np.zeros((2, 2, *theta.shape))
    T[0, 0] = (t1 - t2) * np.cos(theta) ** 2 + t2
    T[1, 1] = (t1 - t2) * np.sin(theta) ** 2 + t2
    T[0, 1] = T[1, 0] = (t1 - t2) * np.sin(theta) * np.cos(theta)

    return T
```

```python
fig, ax = plt.subplots(1, 2, subplot_kw={"projection": "polar"}, figsize=(10, 5))

angles = np.linspace(0, 2 * np.pi, 100)


ax[0].plot(angles, T_ij(1, 0.5, angles)[0, 0], label="Txx")
ax[0].plot(angles, T_ij(1, 0.5, angles)[1, 1], label="Tyy")
ax[0].plot(angles, T_ij(1, 0.5, angles)[0, 1], label="|Txy|")
ax[0].plot(angles, T_ij(1, 0.5, angles)[1, 0], label="|Txy|")

ax[1].plot(angles, -_hopping_matrix(1, 0.5, angles)[0, 0], label="Txx")
ax[1].plot(angles, -_hopping_matrix(1, 0.5, angles)[1, 1], label="Tyy")
ax[1].plot(angles, -_hopping_matrix(1, 0.5, angles)[0, 1], label="|Txy|")
ax[1].plot(angles, -_hopping_matrix(1, 0.5, angles)[1, 0], label="|Txy|")

ax[1].legend()
```
