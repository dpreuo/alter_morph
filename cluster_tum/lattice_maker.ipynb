```python
from alter_morph.lattice_utilities import alter_lattice_maker
from koala.graph_utils import lloyd_relaxation
from koala import plotting as kplt
import pickle
import koala
import matplotlib.pyplot as plt


system_length = 20
lattice_type = "voronoi"
lattice = alter_lattice_maker(system_length, lattice_type)
lattice = lloyd_relaxation(lattice, 200)

lattice = koala.graph_utils.cut_boundaries(lattice, [False, True])

pickle.dump(
    lattice, open(lattice_type + "_" + str(system_length) + "_relaxed.pickle", "wb")
)

fig, ax = plt.subplots(figsize=(10, 10))
kplt.plot_edges(lattice)
```
