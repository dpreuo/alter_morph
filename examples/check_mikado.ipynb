```python
# from alter_morph.mikado_lattice import koala_mikado

from alter_morph.mikado_lattice import random_line_graph, koala_mikado, relaxation
from koala import plotting as pl

import numpy as np
import matplotlib.pyplot as plt
```

```python
lattice = random_line_graph(1, 100)
klattice = koala_mikado(1000, 0)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
plt.sca(ax[0])
plt.scatter(lattice[0][:, 0], lattice[0][:, 1], c="r")
for i, j in lattice[1]:
    plt.plot(
        [lattice[0][i, 0], lattice[0][j, 0]], [lattice[0][i, 1], lattice[0][j, 1]], "k-"
    )
pl.plot_edges(klattice, ax=ax[1])
```

```python
coords, bonds = lattice

fig, ax = plt.subplots(1, 4, figsize=(20, 5))

for n in range(len(ax) - 1):
    for i, j in bonds:
        ax[n].plot([coords[i, 0], coords[j, 0]], [coords[i, 1], coords[j, 1]], "k-")

    # coords2, bonds2 = coords, bonds
    coords, bonds = relaxation(coords, bonds, 1, repeats=2)

for i, j in bonds:
    ax[-1].plot([coords[i, 0], coords[j, 0]], [coords[i, 1], coords[j, 1]], "k-")
```
