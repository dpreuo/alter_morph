```python
from koala.voronization import generate_lattice
from koala.pointsets import uniform
from koala import graph_utils as gu
from koala.lattice import Lattice
from koala.graph_color import color_lattice
from koala import plotting as pl
from tqdm import tqdm


from copy import copy
import numpy as np
import matplotlib.pyplot as plt
```

Some nice demos of the different lattices we know how to make:


```python
# generate a tri-coordinated lattice
number_of_plaquettes = 30
lattice = generate_lattice(uniform(number_of_plaquettes))
```

```python
lattice_2 = gu.make_dual(lattice, reg_steps=0)

pl.plot_edges(lattice)
pl.plot_edges(lattice_2, linewidth=0.3)
```
