```python
import numpy as np
```

```python
sigma_x = np.array([[0, 1], [1, 0]])
sigma_y = np.array([[0, -1j], [1j, 0]])
sigma_z = np.array([[1, 0], [0, -1]])

sigmas = [sigma_x, sigma_y, sigma_z]

s_out = np.zeros((4, 4), dtype=complex)
for i in range(3):
    for j in range(3):
        print(np.kron(sigmas[i], sigmas[j]))
```

```python
print(s_out.real, "\n", s_out.imag)
```
