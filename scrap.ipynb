```python
import numpy as np
from scipy import linalg as la
import adaptive
import matplotlib.pyplot as plt
import itertools
```

```python
sigma_x = np.array([[0, 1], [1, 0]])
sigma_y = np.array([[0, -1j], [1j, 0]])
sigma_z = np.array([[1, 0], [0, -1]])


def func_to_learn(xy):
    x = xy[0]
    y = xy[1]
    m = 1 * sigma_x + 2 * x * sigma_y + 3 * y * sigma_z
    return la.eigvalsh(m)
```

```python
# %%
import numpy as np
import matplotlib.pyplot as plt
from adaptive import Learner1D
from adaptive.runner import simple
from scipy.optimize import root

# %%


# Dummy function to learn, will be replaced later
def func1(a):
    return None


learner = Learner1D(func1, bounds=(1, 10))
initial_guess = 1


# function to "learn" the root of x^4 + 0.5*x^3 - a
def func2(a):
    learner_data = learner.to_numpy()
    if len(learner_data) == 0:  # no previous data
        guess = initial_guess  # needs to be good enough
    else:
        # Find the closest known point to `a` and use that as a guess
        guess = learner_data[np.argmin(np.abs(learner_data[:, 0] - a))][1]
    root_result = root(lambda x: x**4 + 0.5 * x**3 - a, 1)
    return root_result.x[0]


learner.function = func2
# %%
runner = simple(learner, npoints_goal=200)
## %%
data = learner.to_numpy()
plt.plot(data[:, 0], data[:, 1], ".")
plt.xlabel("a")
plt.ylabel("x")
# %%
```
